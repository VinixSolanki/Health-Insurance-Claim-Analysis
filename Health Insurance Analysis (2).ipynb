```python
#Visualize Distribution of Claim Types
claim_type_vals = ins.ClaimType.value_counts()[:10]


ax = (claim_type_vals / ins.shape[0]).plot(kind='bar', color='skyblue')

(claim_type_vals / ins.shape[0]).plot(kind='bar')
plt.title('Top claim types (as % of all)')

ax.bar_label(ax.containers[0], fmt='%.2f%%')
```

```python
ins['ClaimStatus'].value_counts(normalize=True) * 100
```

```python
#Visualize Distribution of Claim Status

categorical_cols = ['ClaimType', 'ClaimSubmissionMethod', 'ProviderSpecialty', 
                    'PatientGender', 'PatientMaritalStatus', 'PatientEmploymentStatus']

for col in categorical_cols:
    ct = pd.crosstab(ins[col], ins['ClaimStatus'])
    ax = ct.plot(kind='bar', stacked=True, figsize=(7,4))
    
    # Add data labels
    for container in ax.containers:
        ax.bar_label(container)
    
    plt.title(f'Claim Status by {col}')
    plt.ylabel('Count')
    plt.show()
```

```python
corr = ins['PatientIncome'].corr(ins['ClaimAmount'])
print(f"Correlation between Income and Claim Amount: {corr:.2f}")
```

```python
bins = [0, 20000, 40000, 60000, 80000, 100000, ins['PatientIncome'].max()]
labels = ['0-20k', '20k-40k', '40k-60k', '60k-80k', '80k-100k', '100k+']

ins['IncomeGroup'] = pd.cut(ins['PatientIncome'], bins=bins, labels=labels)
```

```python
claim_by_income = ins.groupby('IncomeGroup')['ClaimAmount'].sum()  # or .mean() for average
```

```python
plt.figure(figsize=(8,5))
ax = claim_by_income.plot(kind='bar', color='skyblue')
plt.title('Total Claim Amount by Income Group')
plt.xlabel('Income Group')
plt.ylabel('Total Claim Amount')

# Add data labels
for i, v in enumerate(claim_by_income):
    ax.text(i, v + max(claim_by_income)*0.01, f"{v:,.0f}", ha='center')

plt.show()
```

```python
#Visualize Numeric Correlations (Heatmap)
num_cols = ['PatientAge', 'PatientIncome', 'ClaimAmount']
sns.heatmap(ins[num_cols].corr(), annot=True, cmap='Blues')
plt.title('Numeric Feature Correlation')
plt.show()
```
